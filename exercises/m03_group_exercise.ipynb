## Troubleshooting

| Problem | Fix |
|---------|-----|
| `NameError: name 'df' is not defined`, or a `KeyError` right after you switched datasets | Run the setup cell first. After you change the `dataset` line, run the setup cell again and re-run every cell below it in order |
| `ValueError: could not convert string to float` | A text column (such as `Gender`, `Workout_Type`, `Card_Category`) is in your features or in `.corr()`. Use only the columns in `numbers` plus the yes/no column, and use `df.corr(numeric_only=True)` |
| The top bar in the correlation chart is the target with correlation 1.0 | A column always correlates 1.0 with itself. Leave `target` out of the list before you sort |
| `KeyError: 'Session_Duration (hours)'` or a similar column name | Copy the exact name inside quotes. Gym names such as `Session_Duration (hours)` have spaces and brackets |
| `ValueError: Expected 2D array, got 1D array` (newer scikit-learn: `Expected a 2-dimensional container`) | Use double brackets: `df[["Total_Trans_Ct"]]` not `df["Total_Trans_Ct"]` |
| R² is negative | Your features are worse than predicting the mean. Check feature selection and that X and y come from the same rows |
| Task 4 R² is lower than Task 3 | Check that your strongest predictor is one of the four columns, and that both models use `test_size=0.25, random_state=42` |
| The yes/no counts don't match Checkpoint 3 (for example all 0, all 1, or the two counts swapped) | Check the rule and its spelling: `== "Male"` with a capital M for gym, `!= "Blue"` for card. Print the counts with `.value_counts()` |
| RMSE looks like squared units (about 1,500 for gym, millions for card) | Take the square root: `np.sqrt(mean_squared_error(y_test, y_pred))` |
